```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
import numpy as np
```

```python
theorytest = pd.read_csv('car theory test data.csv',encoding='latin-1')
```

```python
theorytest.dtypes
```

```python
theorytest.head(10)
```

```python
test1 = sns.displot(data=theorytest,x='Pass rate (Percentage)',hue='Gender',multiple='dodge')
test1.set(title='UK Theory Test throughout 2007-2012 by genders',ylabel='Frequency')
```

```python
test2 = sns.barplot(data=theorytest,x='Date',y='Pass rate (Percentage)',hue='Gender',ci=None,capsize=0.2)
test2.set(title='UK Theory Test throughout 2007-2012 by genders')
plt.xticks(rotation=90)
```

```python
disabledrive = pd.read_csv('disabled-driving-licence-data.csv')
```

```python
disabledrive.dtypes
```

```python
licence1 = sns.displot(data=disabledrive,x='Year',hue='Age group',multiple='dodge',palette='Paired')
licence1.set(title='UK Disabled Driving Licence via Age Group',ylabel='Frequency')
plt.xticks(rotation=90)
```

```python
licence2= sns.displot(data=disabledrive,x='Year',hue='Disability status',palette='Paired')
licence2.set(title='UK Disabled Driving Licence throughout 2022-2024 via Disability',ylabel='Frequency')
plt.xticks(rotation=90)
```
