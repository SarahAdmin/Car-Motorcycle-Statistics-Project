```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
cars = pd.read_csv('car_data.csv',encoding='latin-1')
```

```python
cars.head(10)
```

```python
make_data = sns.histplot(data=cars,x='Make',hue='Fuel Type',multiple='stack')
make_data.set(title='Make via Fuel Types',ylabel='Number of Cars')
```

```python
trans= sns.histplot(data=cars,x='Make',hue='Transmission',multiple='stack')
trans.set(title='Make via Transmission Types',ylabel='Number of Cars')
```

```python
test = pd.read_csv('theory_test_data.csv',encoding='latin-1')
```

```python
test.info()
```

```python
results = sns.violinplot(data=test,x='Vehicle',y='Score_out_of_50',hue='Status')
results.set(title='Theory Test Results',ylabel='Score')
```

```python
sns.violinplot(data=test,x='Status',y='Score_out_of_50',hue='Status',palette='Set1')
plt.ylabel('Score')
plt.show()
```
