```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
test = pd.read_csv('driving_test_with_zero_faults.csv')
```

```python
test.info()
```

```python
test[' Total car driving tests passed '] = test[' Total car driving tests passed '].replace({',':''},regex=True).astype(int)
```

```python
test[' Total car driving tests passed with zero faults '] = test[' Total car driving tests passed with zero faults '].replace({',':''},regex=True).astype(int)
```

```python
test[' Total car driving tests conducted '] = test[' Total car driving tests conducted '].replace({',':''},regex=True).astype(int)
```

```python
test.dtypes
```

```python
fig, ax = plt.subplots(figsize=(25,10))
sns.barplot(data=test,x=' Total car driving tests passed ',y='Year ',palette='Set1')
for container in ax.containers:
  ax.bar_label(container)
plt.title('Total Car Driving Test Passed 2015-2026')
plt.xlabel('Number of Tests')
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(25,10))
sns.barplot(data=test,x=' Total car driving tests conducted ',y='Year ',palette='Set1')
for container in ax1.containers:
  ax1.bar_label(container)
plt.title('Total Car Driving Test Conducted 2015-2026')
plt.xlabel('Number of Tests')
plt.show()
```

```python
fig, ax2 = plt.subplots(figsize=(25,10))
sns.barplot(data=test,x='Year ',y=' Total car driving tests passed with zero faults ',palette='Set1')
for container in ax2.containers:
  ax2.bar_label(container)
plt.title('Total Car Driving Test Passed with Zero Faults 2015-2026')
plt.xlabel('Number of Tests')
plt.xticks(rotation=75)
plt.show()
```

```python
fig, ax3 = plt.subplots(figsize=(25,10))
sns.lineplot(data=test,x='Year ',y='Car driving test pass rate',marker='o',color='tomato')
for x,y in zip(test['Year '],test['Car driving test pass rate']):
  ax3.text(x,y,str(y),
            horizontalalignment='center',
           verticalalignment='bottom',
           fontsize=12,
           fontweight='bold')
  plt.xticks(rotation=85)
  plt.title('Car Driving Test Pass Rate 2015-2026',fontsize=20)
  plt.ylabel('Pass Rate (%)')
```

```python
fig, ax4 = plt.subplots(figsize=(25,10))
sns.lineplot(data=test,x='Year ',y='Percentage of car driving tests conducted with zero faults',marker='o',color='rebeccapurple')
for x,y in zip(test['Year '],test['Percentage of car driving tests conducted with zero faults']):
  ax4.text(x,y,str(y),
           horizontalalignment='center',
           verticalalignment='bottom',
           fontsize=12,
           fontweight='bold')
plt.xticks(rotation=85)
plt.title('Percentage of Car Drivng Tests Conducted with Zero Faults in 2015-2026',fontsize=20)
plt.ylabel('Percentage (%)')
plt.show()
```

```python
fig, ax5 = plt.subplots(figsize=(25,10))
sns.lineplot(data=test,x='Year ',y='Percentage of car driving test passes with zero faults',marker='o',color='skyblue')
for x,y in zip(test['Year '],test['Percentage of car driving test passes with zero faults']):
  ax5.text(x,y,str(y),
           horizontalalignment='center',
           verticalalignment='bottom',
           fontsize=12,
           fontweight='bold')
plt.xticks(rotation=85)
plt.title('Percentage of Car Drivng Tests Passed with Zero Faults in 2015-2026',fontsize=20)
plt.ylabel('Percentage (%)')
plt.show()
```
