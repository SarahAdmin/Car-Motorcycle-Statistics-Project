```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
theory = pd.read_csv('Driving_instructor_theory_test.csv')
```

```python
theory.info()
```

```python
theory.columns
```

```python
theory.head(10)
```

```python
sns.lineplot(data=theory,x='Date',y='Pass rate (Percentage)',marker='o',color='rebeccapurple')
plt.xticks(rotation=85)
plt.title('Theory Test for Driving Instructors Pass Rate')
plt.xlabel('Year')
plt.show()
```

```python
sns.barplot(data=theory,x=' Passed ',y='Date',palette='Set1',orient='h')
plt.title('Driving Instrctor Passed Theory Test 2007-2027')
plt.xlabel('Year')
plt.show()
```

```python
sns.barplot(data=theory,x=' Conducted ',y='Date',palette='Set1',orient='h')
plt.title('Driving Instrctor Conducted Theory Test 2007-2027')
plt.xlabel('Year')
plt.show()
```
