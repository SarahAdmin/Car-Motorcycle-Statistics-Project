```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
test = pd.read_csv('motorcycle test.csv')
```

```python
test.info()
```

```python
test.head(5)
```

```python
sns.barplot(data=test,x='Passed',y='Date',orient='h',errorbar=None,hue='Gender')
plt.title('Motorcycle Test Passed by Gender 2016-2026')
plt.ylabel('Year')
plt.show()
```

```python
sns.barplot(data=test,x='Conducted',y='Date',hue='Gender',errorbar=None,palette='Set1')
plt.title('Motorcycle Test Conducted by Gender 2016-2026')
plt.ylabel('Year')
plt.show()
```

```python
sns.lineplot(data=test,x='Date',y='Pass rate (Percentage)',marker='o',errorbar=None)
plt.xticks(rotation=80)
plt.title('Motorcycle Test Pass Rate 2016-2026')
plt.show()
```
