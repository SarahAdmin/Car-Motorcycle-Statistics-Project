```python
import pandas as pd
```

```python
import matplotlib.pyplot as plt
```

```python
import seaborn as sns
```

```python
test = pd.read_csv('Driving_Test_data.csv',encoding='latin-1')
```

```python
quarter = pd.read_csv('Quarter_driving_test_data.csv',encoding='latin-1')
```

```python
test.info()
```

```python
test.head(5)
```

```python
test['Passed'] = test['Passed'].replace({',':''},regex=True).astype(int)
```

```python
test['Conducted'] = test['Conducted'].replace({',':''},regex=True).astype(int)
```

```python
test.dtypes
```

```python
sns.barplot(data=test,x='Passed',y='Date',hue='Gender',orient='h',palette='Set1')
plt.title('Driving Test Passed in 2023-2026')
plt.ylabel('Year')
plt.show()
```

```python
sns.barplot(data=test,x='Conducted',y='Date',hue='Gender',orient='h',palette='Set1')
plt.title('Driving Test Conducted in 2023-2026')
plt.ylabel('Year')
plt.show()
```

```python
sns.barplot(data=test,x='Pass rate (Percentage)',y='Date',hue='Gender',orient='h',palette='Set1')
```

```python
quarter.info()
```

```python
quarter['Passed'] = quarter['Passed'].replace({',':''},regex=True).astype(int)
```

```python
quarter['Conducted'] = quarter['Conducted'].replace({',':''},regex=True).astype(int)
```

```python
quarter.dtypes
```

```python
sns.barplot(data=quarter,x='Passed',y='Date',orient='h',hue='Gender')
plt.ylabel('Quarter Year')
plt.title('Quarter Wise Driving Test Passed')
```

```python
sns.barplot(data=quarter,x='Conducted',y='Date',orient='h',hue='Gender')
plt.ylabel('Quarter Year')
plt.title('Quarter Wise Driving Test Conducted')
```

These horizontal bar graphs represents both genders has Passed and Conducted our driving throughout 2023 - 2026.  

```python
sns.histplot(data=quarter,x='Pass rate (Percentage)',hue='Gender',multiple='stack',palette='')
plt.title('Pass Rate')
```

Only Females passed our driving test within 46% and 47% and from 48% and 49% both genders has passed our driving test. Only Males has passed our driving test with 51% pass rate.

```python
sns.boxplot(data=quarter,x='Date',y='Pass rate (Percentage)',palette='bright')
plt.xticks(rotation=85)
plt.xlabel('Quarter Year')
plt.show()
```

```python
sns.violinplot(data=test,x='Date',y='Pass rate (Percentage)',palette='bright')
plt.xlabel('Year')
plt.show()
```

In the violin plot, 2023-2024 was a biggest year for Driving Test Pass Rate. By increasing the number of pass rate from 2023 - 2024 to 2025-2026 is 3-year period.
